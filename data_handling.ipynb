```python
all_file_timelines = {}

# Skipping 3 because its corrupted
for file_num in [1, 2, 4, 5]:
    
    target_header = f"File name: PN00-{file_num}.edf"
    
    start_time = None
    end_time = None
    seizure_start = []
    seizure_end = []
    
    inside_correct_section = False
    
    with open("Seizure-EEG-Data/Seizures-list-PN00.txt", mode="r", encoding="utf-8") as file:
        for line in file:
            
            if target_header in line:
                inside_correct_section = True
                continue
                
            if "File name: " in line and target_header not in line:
                if inside_correct_section:
                    inside_correct_section = False
                    break

            if inside_correct_section:
                if "Registration start time: " in line:
                    start_time = datetime.strptime(line.split("Registration start time: ")[1].strip(), "%H.%M.%S")
                if "Registration end time: " in line:
                    end_time = datetime.strptime(line.split("Registration end time: ")[1].strip(), "%H.%M.%S")
                if "Seizure start time: " in line:
                    seizure_start.append(datetime.strptime(line.split("Seizure start time: ")[1].strip(), "%H.%M.%S"))
                if "Seizure end time: " in line:
                    seizure_end.append(datetime.strptime(line.split("Seizure end time: ")[1].strip(), "%H.%M.%S"))

    seconds_from_start_list = []
    seizure_difference = []
    
    for i in range(len(seizure_start)):
        seconds_from_start = (seizure_start[i] - start_time).total_seconds()
        duration = (seizure_end[i] - seizure_start[i]).total_seconds()
        
        seconds_from_start_list.append(seconds_from_start)
        seizure_difference.append(duration)
    
    all_file_timelines[f"PN00-{file_num}"] = {
        "seizure_seconds_from_start": seconds_from_start_list,
        "seizure_durations_seconds": seizure_difference
    }
print(all_file_timelines)
```

```python
four_sec_chops = {}
all_file_labels = {}
file_numbers = [1, 2, 4, 5]

for filtered_raw, file_num in zip(filtered_raws, file_numbers):
    low = 0
    high = 2048
    eeg_matrix = filtered_raw.get_data()
    total_columns = eeg_matrix.shape[1]
    
    four_sec_chops[f"PN00-{file_num}"] = []
    all_file_labels[f"PN00-{file_num}"] = []
    
    file_key = f"PN00-{file_num}"
    seizure_starts = all_file_timelines[file_key]["seizure_seconds_from_start"]
    seizure_durations = all_file_timelines[file_key]["seizure_durations_seconds"]

    while high <= total_columns:  
        chunk_start_sec = low / 512
        chunk_end_sec = high / 512
        
        four_sec_chops[file_key].append(eeg_matrix[:, low:high])
        
        is_seizure = 0
        
        for start, duration in zip(seizure_starts, seizure_durations):
            seizure_end = start + duration
            
            # Check for overlap
            if chunk_end_sec > start and chunk_start_sec < seizure_end:
                is_seizure = 1
                break  
                
        all_file_labels[file_key].append(is_seizure)
        
        low += 2048
        high += 2048

print("Chunks in File 1:", len(four_sec_chops["PN00-1"]))
print("Labels in File 1:", len(all_file_labels["PN00-1"]))
print("Total seizure chunks marked in File 1:", sum(all_file_labels["PN00-1"]))
print("---")
print("Shape of a single chunk:", four_sec_chops["PN00-1"][0].shape)
```
